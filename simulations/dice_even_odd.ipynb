```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)

numrolls=[]
probeven=[]
probodd=[]

for n in range(1,10001):
    rolls=np.random.randint(1,7,size=n)
    evencount=np.sum(rolls%2==0)
    oddcount=np.sum(rolls%2==1)

    numrolls.append(n)
    probeven.append(evencount/n)
    probodd.append(oddcount/n)

plt.figure(figsize=(8,5))
plt.plot(numrolls,probeven,label="P(even)")
plt.plot(numrolls,probodd,label="P(odd)")
plt.axhline(0.5,linestyle="--",color="black",label="Theoretical probability")

plt.xlabel("Number of rolls")
plt.ylabel("Probability")
plt.title("Convergence of even/odd probabilities for dice rolls")
plt.legend()
plt.grid(True)
plt.savefig("plots/even_odd_probability.png",dpi=300)
plt.show()
```
